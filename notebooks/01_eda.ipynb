```python
import pandas as pd

df=pd.read_csv(
    "../data/GSE2034_series_matrix.txt.gz",
    sep='\t',
    comment='!',
    index_col=0,
    compression='gzip'
)
print(df.shape)
df.head()
```

```python
df_t=df.T
print(df_t.shape)
df_t.head()
```

```python
import gzip
from io import StringIO

with gzip.open("../data/GPL96.annot.gz", 'rt') as f:
    lines = f.readlines()

# Find the line that says the real table starts, and where it ends
start = next(i for i, l in enumerate(lines) if l.startswith('!platform_table_begin')) + 1
end = next(i for i, l in enumerate(lines) if l.startswith('!platform_table_end'))

table_text = ''.join(lines[start:end])
annot = pd.read_csv(StringIO(table_text), sep='\t', low_memory=False)

print(annot.shape)
annot.head()
```

```python
esr1_rows=annot[annot['Gene symbol']=='ESR1']
esr1_rows
```

```python
Y=df_t['205225_at']


X=df_t.drop(columns=['205225_at', '211233_x_at'])

print("Y shape:", Y.shape)
print("X shape:", X.shape)

Y.head()
```

```python
correlations=X.corrwith(Y)
correlations.head()
```

```python
correlations_abs=correlations.abs().sort_values(ascending=False)
correlations_abs.head(20)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
plt.plot(correlations_abs.values[:500])
plt.xlabel('Gene rank')
plt.ylabel('Absolute correlation with ESR1')
plt.title('Correlaion strength drop-off(top 500 genes)')
plt.show()
```

```python
top_genes=correlations_abs.head(100).index
X_selected=X[top_genes]
print(X_selected.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test= train_test_split(
    X_selected, Y, test_size=0.2, random_state=42
)
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("Y_train:", Y_train.shape)
print("Y_test:", Y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error

lr=LinearRegression()
lr.fit(X_train, Y_train)

y_pred_lr=lr.predict(X_test)

print("R²:", r2_score(Y_test, y_pred_lr))
print("RMSE", mean_squared_error(Y_test, y_pred_lr) ** 0.5)
      
```
